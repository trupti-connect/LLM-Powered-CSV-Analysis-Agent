```python
import os
import getpass
import pandas as pd

from langchain_groq import ChatGroq
from langchain_experimental.agents import create_csv_agent
```

```python
os.environ["GROQ_API_KEY"] = getpass.getpass("Enter your Groq API Key: ")
```

```python
CSV_PATH = "House Price India.csv"

df = pd.read_csv(CSV_PATH)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
groq_chat = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0
)
```

```python
agent = create_csv_agent(
    llm=groq_chat,
    path=CSV_PATH,
    verbose=True,
    allow_dangerous_code=True
)
```

```python
response = agent.invoke({
    "input": "How many rows and columns are in this dataset?"
})

print(response["output"])
```

```python
questions = [
    "How many rows and columns are in this dataset?",
    "Display all column names.",
    "Check missing values in each column.",
    "Identify duplicate records.",
    "Generate descriptive statistics.",
    "Find the top 5 frequent values in categorical columns.",
    "Identify outliers in numerical columns.",
    "Calculate correlation between numerical columns.",
    "Identify the strongest relationships between variables.",
    "Generate 5 important business insights.",
    "Recommend suitable visualizations.",
    "Identify important features for machine learning."
]

for i, question in enumerate(questions, 1):

    print(f"\nQuestion {i}: {question}")

    response = agent.invoke({
        "input": question
    })

    print("Answer:", response["output"])
    print("-" * 80)
```
